```python
def min_max(my_list):
    minimum = min(my_list)
    maximum = max(my_list)

    return (minimum,maximum)

min_max([20,30,34,60]) # => (20,60)
```

```python
def double(my_list):
    list = []

    for word in my_list:
        list.append(word*2)

    return list
```

```python
def sum_lists(list_1, list_2):
    sum_list = []
    for i in range(len(list_1)):
        sum_list.append(list_1[i]+list_2[i])

    return sum_list

sum_lists([1,2,3],[3,2,1]) # => [4,4,4]
sum_lists([2,4,6],[3,6,9]) # => [5,10,15]
```

```python
def every_pair(list_1, list_2):

    pair_list = []

    for num1 in list_1:
        for num2 in list_2:
            pair_list.append((num1,num2))

    return pair_list

every_pair([1,2,3],[5,10])
```

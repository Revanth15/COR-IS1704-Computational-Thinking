```python
# Set
#
# Set - an unordered collection of unique items

anson_groceries = ["egg","egg","egg","apple", "apple", "bread", "milk"]

anson_set_of_groceries = set(anson_groceries)
anson_set_of_groceries

tony = {'apple', 'orange', 'pear'}
wife = {'pear', 'banana'}
baby = {}

# What did Tony and Anson buy together?
# Union - This is kind of like addition
tony.union(anson_set_of_groceries)
anson_set_of_groceries.union(tony) # This gives us the same collection

# What did the whole family buy?
tony.union(anson_set_of_groceries).union(wife).union(baby)

# To figure our which items belong to multple sets or collections, we can use 
# intersect
tony.intersection(wife)
wife.intersection(tony) # exact samee result

# to figure our which items only tony likes but not wide, we can use
# difference
tony.difference(wife)
wife.difference(tony)

# union
# intersect
# difference
# is_subset
```

```python
def get_item_v1(my_list, index):
    for i in range(len(my_list)):
        if i == index:
            return my_list[i]

def get_item_v2(my_list, index):
    return my_list[index]
```

```python
def sum_pair_exists(my_set, target):
    for num_1 in my_set:
        for num_2 in my_set:
            if (num_1 + num_2) == target and num_1 != num_2:
                return True
    return False

def sum_pair_exisits_v2(my_list, target):
    for item in my_list:
        complement = target - item
        if complement in my_list and item != complement:
            return True
    return False

print(sum_pair_exists({1,2,3}, 4)) # => True
print(sum_pair_exists({1,2,3}, 5)) # => True
print(sum_pair_exists({1,2,3}, 6)) # => False
```

# Functions
- functions have a name
- they somtimes have parameters ( these are in the function definition, and specify how many arguments a call make with with it)
- they generally return value
- f(x) ( math function for example )
- in some rare cases, a function can also have side-effects. An example of a side-effect is printing something to the screen
- note: printing something is different from the function returning something.

```python
def get_apples(count):
    print("Anson is picking up " + str(count + 1) + " apples")
    return count + 1

apples = get_apples(10)
print("Anson brough back " + str(apples) + " apples")
```

```python
import math

# sqrt(a^2 + b^2)
def hypothenus(a, b):   # this is a function signature
#              ^  ^     # These are called...parameters
    return math.sqrt(a**2 + b**2)


hypothenus(1,1)
#          ^ ^            These are called....arguments
```

```python
# Math function helpers

import math

math.sqrt(16.0) # ==> 4.0 
math.floor(4.1) # ==> 4 -- Rounds down to the closest integar
math.ceil(4.1) # ==> 5 -- rounds up to the closest integer
math.gcd(12, 18) # ==> 6 -- Greatest common denominator
max(3,7,1,9) # ==> 9 -- Largest number
min(3,7,1,9) # ==> 1 -- Smallest number
abs(-7) # ==> 7 absolute value

# you can compose functions
abs(min(-3,-8,1,-9)) # ==> abs(-9) => 9

len("hello!") # ==> 6 -- counts the number of character in the string
```

```python
# Anson needs bags. He needs to bring home the apples and he only has two hands.
# Each bag can hold 3 apples.

def bags_needed(apples):
    return math.ceil(apples / 3)

print("For 3 apples, he needs " + str(bags_needed(3)) + " bags") # ==> 1
print("For 4 apples, he needs " + str(bags_needed(4)) + " bags") # ==> 2
print("For 7 apples, he needs " + str(bags_needed(7)) + " bags") # ==> 3
print("For 1 apples, he needs " + str(bags_needed(1)) + " bags") # ==> 1
```

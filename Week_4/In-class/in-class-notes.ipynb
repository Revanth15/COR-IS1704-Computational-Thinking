```python
len(greeting) # => 5

greeting.upper() # => "HELLO"
                # takes the letters and returns all capitalized

greeting.lower() # => "hello"

"BYE-BYE".lower() # => "bye-bye"
```

```python
# What is a string?
# A string is a sequence of characters

greeting = "hello"

# we can index into a string to get a character
greeting[0] # => 'h'
greeting[1] # => 'e'
# ...
greeting[-1] # => 'o'

# What is the maximum index I can choose without an error?
greeting[4] # => 'o'
# greeting[5] # => IndexError

# In computers, we can start indexing at 0
```

```python
# String Slicing
# A mechanism to part of a sequence rather than a single item in a sequence

greeting = "hello"
greeting[0] # => 'h'

# sequence[start,stop,step]
# - start is the item that you want to start hearing from (inclusive)
# - stop is the item that you want to stop at (not inclusive) (optional)
# - step is the increment that we go at (optional) = 1

greeting
```

```python
# The "in" keyword

word = "hello"

'e' in word # => True
'h' in word # => True
'z' in word # => False
```

```python
# A for-loop and another use of the in keyword
# This for loop does the following: it assigns the first item in the sequence to ch.
# Then it completes the instructions, and
# then assigns ch the next item in the sequence, and continues

word = "hello"

for ch in word:
    print(ch)
```

```python
word = "Hello Tony"

for ch in word:
    print(ch)
```

```python
# Count how many times 'u' appears in the word

word = "cucumber"

u_count = 0

for ch in word:
    if ch == 'u':
        u_count += 1

print(u_count)
```

```python
# range(num)
# Range generates a sequence of numbers from 0...num-1
```

```python
# lets print out only numbers that are even

for i in range(50):
    if i % 2 == 0:
        print(i)
```

```python
# Range has a similar sort of syntax as slicing
# string[start, stop, step]
# range[start, stop, step]

start = 0
stop = 10
step = 2

for i in range(start,stop,step):
    print(i)

# => 0,2,4,6,8
```

```python
greeting = "hello"

for i in range(len(greeting)):
    print(greeting[i])

for ch in greeting:
    print(ch)
```

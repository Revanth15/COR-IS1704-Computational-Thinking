```python
def square(number_list):
    square_list = []
    for number in number_list:
        square = number * number
        square_list = square_list + [square]

    return square_list

square([1, 5, 3])
```

```python
# returns a list of numbers, where each number is doubled

def double(numbers):
    double_list = []
    for number in numbers:
        doubled = number * 2
        double_list = double_list + [doubled]

    return double_list
    
double([1, 5, 3])
```

```python
def word_count(sentence):
    return len(sentence.split(' '))

word_count("Hello! This is a short sentence.") # => 6
```

```python
#readability of text
def difficulty(sentence):
    count=word_count(sentence)
    if count<=11:
        print("Easy")
    elif count<20:
        print("OK")
    elif count<28:
        print("difficult")
    else:
        print("Very difficult")
```

```python
def get_sentences(paragraph):
    return paragraph.split(".")
```

```python
paragraph = ""
sentences = get_sentences(paragraph)
for sentence in sentences:
    difficulty(sentence)
```

Write a program that prompts the user for a 5-digit number, then prints the sum of the digits.
(you may assume at this point of progress of learning that the input has 5 digits)

Output of program:\
   Enter number: 34901\
   3 + 4 + 9 + 0 + 1 = 17

```python
number = input("Enter number: ")

# Experimentation for summation [IGNORE]
total = 0
for i in number:
    total += int(i)


# without loop (uses modulo to determine the remainder)
int_number = int(number)
number_1 = int_number % 10
number_2 = int_number // 10 % 10
number_3 = int_number // 100 % 10
number_4 = int_number // 1000 % 10
number_5 = int_number // 10000 % 10
sum_of_numbers = number_1 + number_2 + number_3 + number_4 + number_5

print(f"{str(number_5)} + {str(number_4)} + {str(number_3)} + {str(number_2)} + {str(number_1)} = {str(sum_of_numbers)}")
```

Write a program that prompts the user for an integer representing a time in seconds, and converts it
to hours, minutes, and seconds, in the format HH:MM:SS. (Hint: f"{hours:02d}" pads a number
to 2 digits with a leading zero)

Output of the program:
    Enter time in seconds: 5000
    5000 seconds is 01:23:20 (HH:MM:SS)

```python
total_seconds = int(input("Enter time in seconds: "))
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60

print(f"{total_seconds} seconds is {hours:02d}:{minutes:02d}:{seconds:02d} (HH:MM:SS)")
```

Write a program that displays the squares and cubes of numbers from 0 to 10, and uses tab
characters (\t) to separate the output into columns. You have not yet learnt loop structures, so you
can write code to calculate the 11 lines of output as follow

N N^2 N^3\
0 0 0\
1 1 1\
2 4 8\
3 9 27\
4 16 64\
5 25 125\
6 36 216\
7 49 343\
8 64 512\
9 81 729\
10 100 1000

```python
print(f"N\tN^2\tN^3")
for i in range(11):
    print(f"{i}\t{i**2}\t{i**3}")
```

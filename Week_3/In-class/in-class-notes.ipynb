```python
# Prepare a function called enough_apples(). It has a parameter apple_count
# It checks whether there are enough, or too few or too many apples for the family
# The family has three people. We want to make sure everyone has atleast one apple, and there are no more than 2 apples per person.
# It should return a boolean: True or False.

def enough_apples(apple_count):
    fam_pax = 3
    num_apples_per_pax = apple_count // fam_pax
    if num_apples_per_pax > 2 or num_apples_per_pax == 0:
        return False
    else: 
        return True

enough_apples(3)
```

```python
# Tuples
# Tuples are a datatype that you can create on the fly
# They represent a bundle of information.
# 
# gps = (lat, long)
# birthday = (year, month, day)
# student = ('Benjamin', 'Turner', 18)

# This funcitons returns a tuple -> (area, perimeter)
def rectangle_info(length, width):
    area = length * width
    perimeter = 2 * (length + width)
    return (area, perimeter)

area, perimeter = rectangle_info(3,4) # you can unpack the tuple right away
print(area)
```
